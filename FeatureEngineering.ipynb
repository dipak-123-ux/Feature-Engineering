5. 	What are continuous and categorical variables?   
    -	Continuous Variables
               Continuous variables represent quantities that you measure. Because they     exist on a continuous spectrum, there is always another possible value between any     two given points.
Example: A person's height could be 175 cm, 175.5 cm, or 175.523 cm depending on how precise your measuring tool is.
In ML Pipelines: These variables are usually fed directly into algorithms, but they often require feature scaling (like Standardization or Min-Max Scaling) so that columns with massive numbers don't overpower columns with smaller numbers.
Categorical Variables
Categorical variables represent characteristics or qualities that you count or sort. They place data points into mutually exclusive buckets.
Example: A variable named Device_Type might only contain the values ["Mobile", "Tablet", "Desktop"].
In ML Pipelines: Most machine learning algorithms cannot understand raw text or categories. Therefore, these variables must undergo categorical encoding (like One-Hot Encoding or Ordinal Encoding) to turn the text buckets into numeric arrays (0s and 1s) before training.

15. 	What is causation? Explain difference between correlation and causation with an example.
    -	Causation (or causality) means that one event or variable directly produces or causes a change in another. It implies a direct cause-and-effect relationship: action A explicitly makes outcome B happen.
    While correlation tells us that two variables change together, causation explains why they change together by proving that one is driving the other.
    A Clear Example: The Fire Truck Paradox
    Imagine looking at historical emergency data for a city's fires:
    The Correlation: Data shows that the more fire trucks that show up to a        house fire, the worse the property damage turns out to be. There is a    strong, undeniable positive correlation between fire trucks and damage.
                                                     The Flawed Conclusion (Causation Mistake): If you assumed correlation equals causation, you would conclude that sending fire trucks to a scene causes houses to burn down, and therefore the city should stop dispatching fire trucks to reduce damage.
   The Reality (The Hidden Cause): A hidden third variable—the size of the fire (the confounding variable)—is driving both. A massive fire causes the city to deploy more fire trucks, and that same massive fire causes severe property damage. The trucks themselves do not cause the damage.

[ Traditional Gradient Descent ] ──► [ Momentum-Based ] ──► [ Adaptive Learning Rate ] • Batch GD • SGD with Momentum • AdaGrad / RMSprop • Stochastic GD (SGD) • Adam

Stochastic Gradient Descent (SGD)	Updates the model's weights immediately after looking at each individual data row.	Large datasets; introduces randomness that helps escape local pitfalls.	The Panic Hiker: A person lost on a foggy mountain who takes a step, recalculates their direction immediately, and changes paths every single second.
Mini-Batch SGD	A hybrid approach. It splits the data into small chunks (e.g., batches of 32 or 64) and updates weights after each chunk.	Standard default for training deep neural networks.	The Grocery Shopper: Instead of inspecting every item in the store or checking out one item at a time, you fill a basket, check out, and then repeat.

Smoothing out bouncy, erratic weight updates in tricky mathematical terrains.	The Heavy Bowling Ball: Rolling a ball down a bumpy hill. RMSprop	An adaptive learning rate method that restricts updates to a moving average, preventing the learning speed from grinding to a halt.	Recurrent Neural Networks (RNNs) and time-series data.	The Smart Braking System: An autonomous car that slams on the brakes over massive rocks but speeds up on predictable asphalt.

20. 	What are continuous and categorical variables?
    -	Continuous Variables
A continuous variable is a numeric variable that can take on an infinite number of values between any two points. They represent measurements rather than counts.
Examples:
         House Price: ₹4,500,000.50, ₹12,000,000.00
Categorical Variables
A categorical variable represents qualitative data divided into distinct groups or categories. These categories are typically based on a specific trait or characteristic.
Categorical data is further split into two distinct sub-types:
Nominal Variables (No Order): Categories with no inherent ranking or logical sequence.
Examples: Eye color (Blue, Brown, Green), City (Lucknow, Mumbai, Delhi)
Ordinal Variables (Ordered): Categories that follow a clear, meaningful order or scale, but the exact mathematical distance between the ranks is unknown.
Examples: Customer Satisfaction (Low, Medium, High), Education Level (High School, Bachelors, Masters, PhD), Shirt Size (S, M, L, XL).

21. 	What is feature scaling? How does it help in Machine Learning?
    -	Feature scaling is a data preprocessing step used to standardise or normalise the range of independent variables (features) in a dataset.
Speeds Up Gradient Descent
Algorithms that use gradient descent to find the optimal weights (like Linear Regression, Logistic Regression, and Neural Networks) converge much faster with scaled features.
Without scaling: The gradient descent path bounces back and forth in a zig-zag pattern because a small change in a large-scale feature impacts the error function drastically.
With scaling: The error surface becomes symmetric and spherical, allowing the algorithm to take a direct, rapid path straight to the global minimum.
